```python
#Universal token explorer
from nuscenes.nuscenes import NuScenes
from pprint import pprint

# List of all tables in nuScenes that contain tokens
ALL_TABLES = [
    'sample',
    'sample_data',
    'sample_annotation',
    'scene',
    'log',
    'category',
    'attribute',
    'instance',
    'ego_pose',
    'calibrated_sensor',
    'sensor',
    'visibility'
]

def find_token(nusc, token):
    found = False
    for table_name in ALL_TABLES:
        table = getattr(nusc, table_name)
        for record in table:
            if record['token'] == token:
                print(f"\n=== Match found in table: '{table_name}' ===")
                pprint(record)
                found = True
                break
        if found:
            break
    if not found:
        print(f"\n Token not found in any known nuScenes table.")

# ==== Main ====
if __name__ == "__main__":
    
    dataroot = '/data/sets/nuscenes'
    version = "v1.0-trainval"  # or "v1.0-mini", etc.

    
    # any token (sample, annotation, sensor, etc.)
    token = "8fefc430cbfa4c2191978c0df302eb98"
    find_token(nusc, token)
    
```
