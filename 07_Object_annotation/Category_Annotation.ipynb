```python
%matplotlib inline
from nuscenes.nuscenes import NuScenes
from collections import Counter

nusc = NuScenes(version='v1.0-mini', dataroot='/data/sets/nuscenes', verbose=True)
```

```python
# Get a list of all category names
category_names = [cat['name'] for cat in nusc.category]

print("Total Categories:", len(category_names))
print("List of Categories:")
for name in category_names:
    print("-", name)
```

```python
from nuscenes.nuscenes import NuScenes
from collections import Counter

def show_sample_category_stats(nusc, sample_token):
    """
    Prints:
    - Number of annotations in the sample.
    - Categories and counts in the sample.
    - Most and least frequent categories in the sample.
    """
    try:
        sample = nusc.get('sample', sample_token)
    except:
        print(f" Sample token '{sample_token}' not found.")
        return

    print(f"\n Sample Token: {sample_token}")
    print(f" Number of annotations in the sample: {len(sample['anns'])}")

    # Count categories in the sample
    category_counter = Counter()
    for ann_token in sample['anns']:
        annotation = nusc.get('sample_annotation', ann_token)
        instance = nusc.get('instance', annotation['instance_token'])
        category_token = instance['category_token']
        category_name = nusc.get('category', category_token)['name']
        category_counter[category_name] += 1

    print("\n Categories in this sample:")
    for name, count in sorted(category_counter.items()):
        print(f"- {name}: {count} time(s)")

    if category_counter:
        # Most frequent category
        max_cat, max_count = category_counter.most_common(1)[0]
        print(f"\n Most frequent category in sample: {max_cat} ({max_count} times)")

        # Least frequent category(s)
        min_count = min(category_counter.values())
        min_categories = [cat for cat, cnt in category_counter.items() if cnt == min_count]
        print(f"\n Least frequent category(s) in sample:")
        for cat in min_categories:
            print(f"- {cat} ({min_count} time(s))")


sample_token = 'd0ecf474e0e64950aa265cd44b9c9d75'
show_sample_category_stats(nusc, sample_token)
```
