```python
from nuscenes.nuscenes import NuScenes
from pprint import pprint
import os

# IMPORTANT: Use the full dataset path, NOT the _meta version!
dataroot = r"C:\Users\$8CUJ00-BGKRQC8B3RAL\Downloads\data introduce\nuscenes\v1.0-trainval_meta"  # <-- Full dataset path (remove _meta)
version = "v1.0-trainval"

def find_and_visualize_annotation(nusc: NuScenes, token: str):
    # Search only in sample_annotation table
    for ann in nusc.sample_annotation:
        if ann['token'] == token:
            print("\n=== Found annotation ===")
            pprint(ann)

            print(f"\nRendering annotation: {token}")
            nusc.render_annotation(token)
            print("Image with annotation displayed. Close window to continue.")
            return
    print("Annotation token not found.")

if __name__ == "__main__":
    if not os.path.exists(dataroot):
        print(f"Error: Dataroot path does not exist: {dataroot}")
        print("Please update 'dataroot' to your actual nuScenes dataset full data path.")
    else:
        nusc = NuScenes(version=version, dataroot=dataroot, verbose=True)
        token = "6fdf337ff7a640eaadcb3001e3d47f25"  # your annotation token here
        find_and_visualize_annotation(nusc, token)
```
